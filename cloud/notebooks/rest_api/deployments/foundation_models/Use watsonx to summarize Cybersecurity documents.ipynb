### Inferencing class

This cell defines a class that makes a REST API call to the watsonx Foundation Model
inferencing API that we will use to generate output from the provided input.
The class takes the access token created in the previous step, and uses it to
make a REST API call with input, model id and model parameters. The response
from the API call is returned as the cell output.


**Action:** Provide watsonx.ai Runtime url to work with watsonx.ai.


````python
endpoint_url = input("Please enter your watsonx.ai Runtime endpoint url (hit enter): ")
````

Define a `PromptClient` class for prompts generation.


````python
class PromptClient:
    def __init__(self, access_token: str, project_id: str, endpoint_url: str):
        self.project_id = project_id
        self.url = f"{endpoint_url.rstrip('/')}/ml/v1/text/chat"
        self.headers = {
            "Authorization": f"Bearer {access_token}",
            "Content-Type": "application/json",
        }

    def chat(self, model_id: str, messages: list[str], **params):
        payload = {
            "model_id": model_id,
            "messages": messages,
            "project_id": self.project_id,
            **params,
        }

        response = httpx.post(
            self.url,
            params={"version": "2024-03-19"},
            json=payload,
            headers=self.headers,
            timeout=30,
        )
        if response.status_code == 200:
            return response.json()
        else:
            raise RuntimeError(response.text)
````

## Generate document summary


Define instructions for the model.


````python
instruction = """Extract the key outline of the "Original text" similar to the Simplification according to the examples."""
````

Prepare model inputs for zero-shot example, use below zero_shot_inputs.


````python
zero_shot_inputs = [text for text in data_test["Paragraph"]]
for i in range(2):
    print(f"The sentence example {i+1} is:\n {zero_shot_inputs[i]}\n")
````

Prepare model inputs for few-shot examples, use below few_shot_inputs.


````python
data_train_and_labels = data_train.copy()
data_train_and_labels["Simplification"] = y_train
````

````python
train_samples = data_train_and_labels.sample(2)
few_shot_example = []
examples = []
for s in range(len(train_samples)):
    examples.append(
        f"\tsentence:\t{train_samples['Paragraph'].iloc[s]}\n\tSimplification: {train_samples['Simplification'].iloc[s]}\n"
    )
few_shot_examples = ["".join(examples)]
````

````python
few_shot_inputs_ = [text for text in data_test["Paragraph"].values]
for i in range(2):
    print(f"The sentence example {i+1} is:\n {few_shot_inputs_[i]}\n")
````

### Generate the cybersecurity: SPEC5G Cellular Network Protocol summary using `watsonx.ai` model.

**Note:** You might need to adjust model parameters for different models or tasks, to do so please refer to <a href="https://cloud.ibm.com/apidocs/watsonx-ai#text-chat" target="_blank" rel="Infer text params">documentation</a>.


Initialize the `PromptClient` class.

**Hint:** Your authentication token might expire, if so please regenerate the `access_token` and reinitialize the `PromptClient` class.


````python
prompt_client = PromptClient(access_token, project_id, endpoint_url)
````

Get the docs summaries.


````python
results = []

for inp in few_shot_inputs_[:2]:
    results.append(
        prompt_client.chat(
            messages=[
                {"role": "system", "content": instruction + few_shot_examples[0]},
                {"role": "user", "content": inp},
            ],
            model_id=model_id,
        )
    )
````

Explore model output.


````python
print(json.dumps(results, indent=2))
````
